***Wordnet Lemmatizer***

```python
import nltk
nltk.download('wordnet')
```

```python
'''
Lemmatization is similar to stemming but it brings context to the words.
So it links words with similar meaning to one word.
POS TAGGING-> Parts of Speech
Noun->n
Verb->v
Adjective->a
Adverb->r
'''
from nltk.stem import WordNetLemmatizer
wnl=WordNetLemmatizer()
print(wnl.lemmatize('going',pos='v'))
print(wnl.lemmatize('going',pos='n'))
print(wnl.lemmatize('going',pos='a'))
print(wnl.lemmatize('going',pos='r'))
```

```python
# Re-tokenize the paragraph into a list of sentences, as 'sentences' is currently a string.
sentences = nltk.sent_tokenize(paragraph)
for i in range(len(sentences)):
  words=nltk.word_tokenize(sentences[i])
  words=[stemmer.stem(word) for word in words if word not in set(nltk_stopwords.words('english'))]
  sentences[i]=' '.join(words)
```

```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter
# Set the path to the file you'd like to load
file_path = "spam.csv"
# Load the latest version
messages = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "uciml/sms-spam-collection-dataset",
  file_path,
  # Provide any additional arguments like
  # sql_query or pandas_kwargs. See the
  # documenation for more information:
  # https://github.com/Kaggle/kagglehub/blob/main/README.md#kaggledatasetadapterpandas
  pandas_kwargs={'encoding': 'latin1'}
)
print("First 5 records:", messages.head())
```

```python
messages.head()
```

```python
##Data cleaning and preprocessing
```

```python
import re
import nltk
nltk.download('stopwords')
import re
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
  review=re.sub('[^a-zA-Z]',' ',messages['v2'][i])
  review=review.lower()
  review=review.split()
  review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
  review=' '.join(review)
  corpus.append(review)
```

```python
## create the bag words
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(max_features=2500,binary=True) #binary true for binary bag of words #by default uni grams
X=cv.fit_transform(corpus).toarray()
```

```python
X.shape
```
